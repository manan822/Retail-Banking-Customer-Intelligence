# Business Key Performance Indicators (KPIs)

Key Performance Indicators (KPIs) are measurable metrics that help banks monitor business performance, evaluate customer relationships, measure product adoption, and support strategic decision-making. In this project, KPIs have been selected to provide meaningful insights into customer behavior, branch performance, financial health, and product opportunities.

The KPIs are categorized into Customer KPIs, Financial KPIs, Product KPIs, Relationship KPIs, and Branch KPIs.

---

## 1. Customer KPIs

These KPIs help understand the overall customer base and identify high-value customer segments.

### Total Customers
- Definition: Total number of customers available in the dataset.
- Business Importance:
  - Indicates the size of the customer base.
  - Helps estimate the scale of banking operations.

---

### Average Customer Age
- Definition: Average age of all customers.
- Business Importance:
  - Helps identify the dominant customer age group.
  - Assists in designing age-specific banking products.

---

### Average Income
- Definition: Average annual income of customers.
- Business Importance:
  - Measures customers' purchasing power.
  - Helps determine eligibility for premium banking products.

---

### Customer Segment Distribution
- Definition: Percentage of customers belonging to each customer segment.
- Business Importance:
  - Helps understand the composition of the customer base.
  - Supports targeted marketing strategies.

---

## 2. Financial KPIs

Financial KPIs evaluate customers' financial strength and long-term value.

### Average Customer Lifetime Value (CLV)

Definition:
The average estimated lifetime value generated by customers throughout their relationship with the bank.

Business Importance:
- Identifies the overall profitability of customers.
- Helps prioritize customer retention.
- Supports premium customer identification.

---

### Average Financial Health Score

Definition:
Average financial stability score of customers.

Business Importance:
- Measures overall financial wellness.
- Helps identify financially healthy customers.
- Supports personalized banking recommendations.

---

### Average Mortgage

Definition:
Average mortgage amount held by customers.

Business Importance:
- Indicates customer liabilities.
- Helps identify opportunities for refinancing and home loan products.

---

### Average Credit Card Spending (CCAvg)

Definition:
Average monthly spending through credit cards.

Business Importance:
- Measures customer spending behavior.
- Helps identify premium credit card opportunities.
- Indicates transaction activity.

---

## 3. Product KPIs

These KPIs measure the adoption of banking products and services.

### Credit Card Penetration

Definition:
Percentage of customers who own a credit card.

Business Importance:
- Measures product adoption.
- Identifies opportunities for card upgrades.

---

### Online Banking Adoption

Definition:
Percentage of customers using digital banking services.

Business Importance:
- Indicates digital transformation.
- Measures customer engagement with online banking.

---

### Securities Account Penetration

Definition:
Percentage of customers holding a securities account.

Business Importance:
- Measures investment product adoption.
- Helps identify wealth management opportunities.

---

### CD Account Penetration

Definition:
Percentage of customers holding Certificate of Deposit (CD) accounts.

Business Importance:
- Indicates investment behavior.
- Measures adoption of fixed-income products.

---

### Personal Loan Acceptance Rate

Definition:
Percentage of customers who have accepted personal loans.

Business Importance:
- Measures loan product performance.
- Helps evaluate lending strategies.

---

## 4. Relationship KPIs

These KPIs evaluate customer loyalty and relationship strength.

### Average Relationship Years

Definition:
Average duration customers have maintained their relationship with the bank.

Business Importance:
- Indicates customer loyalty.
- Helps measure customer retention.

---

### Average Relationship Priority Score

Definition:
Average priority score assigned to customers based on their business value.

Business Importance:
- Helps relationship managers prioritize customer engagement.
- Identifies customers requiring immediate attention.

---

### High Priority Customers

Definition:
Number of customers with Relationship Priority Score greater than 80.

Business Importance:
- Identifies premium customers.
- Supports personalized relationship management.

---

## 5. Branch KPIs

Branch KPIs evaluate branch-level business performance.

### Customers per Branch

Definition:
Total number of customers served by each branch.

Business Importance:
- Measures branch workload.
- Supports resource allocation.

---

### Average CLV per Branch

Definition:
Average Customer Lifetime Value generated by customers of each branch.

Business Importance:
- Identifies high-performing branches.
- Measures branch profitability.

---

### Average Financial Health Score per Branch

Definition:
Average financial health score of customers in each branch.

Business Importance:
- Indicates customer quality across branches.
- Supports regional business planning.

---

### Product Recommendation Distribution

Definition:
Distribution of the "Next Best Product" recommendations.

Business Importance:
- Helps understand customer needs.
- Supports targeted cross-selling campaigns.

---

# Bivariate Analysis

The objective of bivariate analysis is to understand the relationship between two variables and identify patterns that support business decision-making in retail banking.

## Income vs Customer Lifetime Value (CLV)

### Business Question

Does higher customer income lead to higher Customer Lifetime Value (CLV)?

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='Income', y='Customer Lifetime Value (CLV)')

plt.title("Income vs Customer Lifetime Value")
plt.xlabel("Annual Income (₹ Lakhs)")
plt.ylabel("Customer Lifetime Value")

plt.show()
```

```python
correlation = df[['Income','Customer Lifetime Value (CLV)']].corr().iloc[0,1]

print(f"Correlation between Income and CLV: {correlation:.2f}")
```

The strong positive correlation of 0.86 confirms that customer income is a primary driver of long-term profitability for the bank.
This justifies prioritizing high-income segments for premium retention programs, as they represent the highest potential for lifetime revenue growth.

**Correlation Heatmap**

```python
plt.figure(figsize=(12,8))

correlation_matrix = df.select_dtypes(include=['int64', 'float64']).corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")

plt.show()
```

The correlation heatmap provides a comprehensive overview of linear relationships between numerical variables. Key insights include:

1.  **Strong Positive Correlations with CLV**: `Income` (0.86), `Relationship Priority Score` (0.84), and `Financial Health Score` (0.83) show very strong positive correlations with `Customer Lifetime Value (CLV)`. This confirms that wealthier, financially stable, and highly prioritized customers are the most valuable to the bank, justifying targeted retention and growth strategies for these segments.

2.  **Inter-correlations among Key Metrics**: `Income` is also strongly correlated with `Financial Health Score` (0.77) and `Relationship Priority Score` (0.85). This suggests a virtuous cycle where higher income leads to better financial health, which in turn elevates a customer's priority for the bank. This validates the bank's current customer segmentation and prioritization models.

3.  **Credit Card Spending (`CCAvg`)**: `CCAvg` has a strong positive correlation with `CLV` (0.74), `Income` (0.65), `Financial Health Score` (0.56), and `Relationship Priority Score` (0.56). This indicates that higher credit card spending is associated with overall customer value and financial well-being, highlighting credit card usage as a key indicator of customer engagement and profitability.

4.  **Age and Experience**: `Age` and `Experience` are very highly correlated (0.99), as expected, since experience is typically accumulated with age. Their correlations with other variables are generally weaker compared to income-related metrics.

5.  **Weak Correlations**: Variables like `ID`, `ZIP Code`, `Family`, `Education`, and `Mortgage` show relatively weaker correlations with the primary value drivers like `CLV` and `Income`, suggesting they might be less direct indicators of customer profitability compared to behavioral or financial health metrics.

### Business Implications:

-   **Integrated Product Strategy**: Campaigns that promote products influencing these highly correlated variables (e.g., wealth management for high-income/high-FHS clients, premium credit cards for high-CCAvg users) are likely to yield the best results.
